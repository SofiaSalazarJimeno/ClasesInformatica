```python

# EL MISTERIO DEL VAGÓN 7
# Programa interactivo basado en el diagrama de flujo del trabajo.

# ---------- INICIO: presentación de la historia ----------

import time
import builtins

# Guardamos el print original de Python
print_original = builtins.print

# Creamos un print que muestra el texto línea a línea
def print(*args, **kwargs):
    texto = " ".join(str(arg) for arg in args)

    for linea in texto.split("\n"):
        print_original(linea, **kwargs)
        time.sleep(1)

print("=" * 55)
print("           EL MISTERIO DEL VAGÓN 7")
print("=" * 55)
print("Una noche de tormenta, el tren El Horizonte Europeo viaja de París")
print("a Madrid. Eres un antiguo detective que viaja para descansar...")
print("hasta que Eduardo desaparece.")
print()

# ---------- Lista de personajes ----------
print("PERSONAJES:")
print("- Eduardo: coleccionista de arte (desaparecido)")
print("- Ana: joven escritora que lo observa todo")
print("- Tomás: médico con un maletín negro")
print("- Víctor: empresario rico y arrogante")
print("- Margarita: mujer mayor, distraída")
print("- Luis: camarero del tren")
print("- Sergio: revisor del vagón 7")
print("- Elena: mujer elegante que viaja sola")
print("- Eva: hermana gemela de Elena")
print()

# ---------- Eduardo desaparece ----------
print("A las 23:20 se apagan las luces de todo el tren.")
print("Cuando vuelven, Eduardo ya no está en su compartimento.")
print("Las vías están bloqueadas. Nadie puede salir.")
print("¡Empieza la investigación!")
print()

#Crear variable con nombre de jugador
nombre = input("¿Cómo te llamas, detective? ")
print("Bienvenida/o,", nombre, "- el caso es tuyo.")
print()

# ---------- Variables del juego ----------
acciones = 0
pistas = []

# ---------- Función para añadir pistas ---------- (append añade elemento al final de la lista)
def anadir_pista(texto):
    if texto not in pistas:
        pistas.append(texto)

# ---------- Bucle principal ----------
while True:
    print("-" * 55)
    print("¿Qué quieres hacer?")
    print("1. Interrogar sospechosos")
    print("2. Examinar el compartimento de Eduardo")
    print("3. Revisar huellas (zapatos)")
    print("4. Investigar la fotografía")
    print("5. Buscar a Elena")
    print("6. Analizar pistas")
    print("7. Acusar al culpable")

#el .strip lo usamos para eliminar lo que haya antes en la cadena de texto (por si alguien escribe primero 3 espacios y luego 6 y asi se ejecute la ultima seleccionada)
    opcion = input("Elige una opción (1-7): ").strip()
    print()

    acciones = acciones + 1

    # ----- Opción 1: interrogar sospechosos -----
    if opcion == "1":
        print("¿A quién quieres interrogar?")
        print("Víctor, Tomás, Ana, Margarita, Luis, Sergio, Elena")

#.strip elimina lo anterior y el .lower lo convierte en minusculas
        personaje = input("Escribe el nombre: ").strip().lower()
        print()

        if personaje == "víctor" or personaje == "victor":
            print("Víctor: 'Estaba en mi compartimento.")
            print("Nadie puede confirmarlo.'")
            print("Reconoce que discutió con Eduardo por un negocio,")
            print("pero no da más detalles.")

        elif personaje == "tomás" or personaje == "tomas":
            print("Tomás: 'Estaba leyendo en mi compartimento.'")

        elif personaje == "ana":
            print("Ana: 'Estaba escribiendo en mi libreta.'")

        elif personaje == "margarita":
            print("Margarita: 'Estuve sentada en el pasillo.'")

        elif personaje == "luis":
            print("Luis: 'Estaba buscando unas velas.'")

        elif personaje == "sergio":
            print("Sergio: 'No me moví del extremo del vagón")
            print("durante el apagón.'")
            print("Tú: '¿Estás seguro?'")
            print("Sergio: 'Completamente.'")

        elif personaje == "elena":
            print("Elena: 'No conocía a Eduardo.")
            print("Me pareció desagradable.'")
            print("Dice que nunca habló con él.")
            print("Sospechas que miente.")

        else:
            print("Ese personaje no existe.")

    # ----- Opción 2: examinar el compartimento -----
    elif opcion == "2":
        print("Examinas el compartimento de Eduardo y encuentras:")
        print("- Una copa rota en el suelo")
        print("- Un reloj de bolsillo detenido a las 23:17")
        print("- Un guante negro junto a la cama")
        print("- Una llave escondida dentro de un libro")
        print("- Una carta sin firma:")
        print("'Si vuelves a hablar de aquella noche,")
        print("todos sabrán lo que hiciste.'")

        anadir_pista("Reloj parado a las 23:17")
        anadir_pista("Guante negro junto a la cama")
        anadir_pista("Llave escondida en un libro")
        anadir_pista("Carta de chantaje sin firma")

    # ----- Opción 3: revisar huellas -----
    elif opcion == "3":
        print("Compruebas los zapatos de todos:")
        print("- Víctor: limpios")
        print("- Tomás: limpios")
        print("- Ana: limpios")
        print("- Margarita: limpios")
        print("- Elena: impecables")
        print("- Sergio: ¡TIENEN BARRO!")
        print("Hay una huella de barro junto a la puerta de servicio.")

        anadir_pista("Solo Sergio tiene barro en los zapatos")

    # ----- Opción 4: investigar la fotografía -----
    elif opcion == "4":
        print("Entre las cosas de Eduardo hay una fotografía antigua.")
        print("Detrás hay una fecha: 2016, y dos nombres:")
        print("'Elena' y 'Eva'.")
        print("No sabes quién es Eva...")

        anadir_pista("Fotografía de 2016 con los nombres Elena y Eva")

    # ----- Opción 5: buscar a Elena -----
    elif opcion == "5":
        print("Vas al compartimento de Elena:")
        print("está vacío y la maleta ha desaparecido.")
        print("Por el pasillo aparece una mujer que parece Elena.")
        print("Le enseñas la fotografía y pregunta:")
        print("'¿Quién es Eva?'")
        print("Ella responde: 'Mi hermana.'")
        print("Entonces aparece OTRA mujer idéntica.")
        print("¡Son gemelas! Por eso todos veían a Elena")
        print("en sitios distintos y su comportamiento parecía cambiar.")
        print("Las hermanas cuentan que buscaban pruebas de que Eduardo")
        print("estaba metido en una red de falsificación de obras de arte.")
        print("Pero no saben quién se llevó a Eduardo.")

        anadir_pista("Elena y Eva son hermanas gemelas")

    # ----- Opción 6: analizar pistas -----
    elif opcion == "6":
        print("Relacionas todas las pistas:")
        print("- El reloj marca las 23:17, pero el apagón")
        print("  empezó a las 23:20.")
        print("  Alguien manipuló el reloj para falsear la hora.")
        print("- La llave abre la puerta de servicio.")
        print("- Hay barro junto a esa puerta y barro")
        print("  en los zapatos de Sergio.")
        print("- Sergio dijo que no se movió de su puesto.")
        print("- Solo Sergio conoce a la perfección")
        print("  el funcionamiento del tren.")
        print()
        print("Pistas descubiertas hasta ahora:")

        if len(pistas) == 0:
            print("(todavía ninguna, investiga más)")

        for pista in pistas:
            print("-", pista)

    # ----- Opción 7: acusar al culpable -----
    elif opcion == "7":
        print("¡Es hora de acusar al culpable!")
        print()
        print("SOSPECHOSOS: Víctor, Tomás, Ana, Margarita, Luis, Sergio, Elena")

        respuesta = input("¿Quién crees que secuestró a Eduardo? ").strip().lower()

        print()

        if respuesta in ["sergio"]:
            print("¡CORRECTO! Sergio es el culpable.")
            print()
            print("Explicación final:")
            print("Sergio también participaba en la falsificación de arte")
            print("y Eduardo lo chantajeaba.")
            print("Aprovechó el apagón para secuestrarlo y lo escondió")
            print("en un compartimento de la puerta de servicio.")
            print("Manipuló el reloj (23:17) para falsear la hora")
            print("y dejó pistas para que sospecharan de Elena,")
            print("sabiendo que las gemelas causarían confusión.")
            print("Pero dejó barro junto a la puerta de servicio.")
            print("Eduardo fue encontrado con vida y Sergio fue detenido.")
            print()
            print("Has resuelto el caso en", acciones, "acciones,", nombre, ".")
            print("FIN. Gracias por jugar a 'El misterio del vagón 7'.")
            break

        else:
            print("Incorrecto. Revisa las pistas y sigue investigando.")
            print()
            continue

    # ----- Opción no válida -----
    else:
        print("Opción no válida. Escribe un número del 1 al 7.")
        continue

    # ---------- ¿Quieres hacer otra acción? ----------
    print()
    otra = input("¿Quieres hacer otra acción? (si/no): ").strip().lower()

    if otra not in ["si", "sí", "s"]:
        print()
        print("Puedes volver a jugar cuando quieras.")
        print("FIN DEL JUEGO.")
        break
```
